# Using Machine Learning Algorithms to Determine the Appropriateness of Management's Going Concern Assumption

## Introduction
### The Financial Statement Audit
A financial statement audit (here after “audit”) is an independent review of an entity’s records to determine the extent to which the entity’s financial statements and non-financial disclosures represent a true and fair view of the entity. The audit is a risk-based approach. The auditor starts by reviewing information in the financial statements. If the review outlines adverse trends or inconsistencies, the auditor may expand the scope of the audit (Westland, 2020).

### The Going Concern Assumption
Under the going concern basis of accounting, the financial statements are prepared with the assumption that an entity is a “going concern” and will continue its operations for the foreseeable future (IAASB, 2015). This distinction is important. When the going concern basis of accounting is not appropriate for the entity, an alternative model such as the liquidation basis of accounting must be applied. 

There are valuation differences to consider under differing basis of accounting. When financial statements are prepared under the going concern basis of accounting, assets and liabilities are valued on the basis that they will be settled during the ordinary course of business. For entities using the liquidation basis of accounting this approach may be inappropriate since the process of liquidation may involve forced or distressed sales. Logically, the values for distressed asset and liability settlement may deviate from the values recorded when the entity was a going concern. Correspondingly adjustments to financial statement values may need to be applied (PricewaterhouseCoopers, 2021). 

### Auditing the Going Concern Assumption
Most if not all professional standards for financial statement audits require the auditor to assess the appropriateness of the going concern basis of accounting. International Standards on Auditing (“ISAs”) require the auditor to “obtain sufficient and appropriate audit evidence regarding, and conclude on, the appropriateness of the audited entity’s use of the going concern basis of accounting” (IAASB, 2015). 

Auditors are facing increased scrutiny in the wake of recent corporate collapses. In the UK, the value of fines levied by the Financial Reporting Council to auditing firms had risen to £33.3m in the year leading up to September 30, 2022. This represents an increase of 77 percent over the previous year (The Accountant Online, 2022). The Accountant Online further reports that a large portion of this increase is attributed to a fine of £14.4m to the auditors of Carillion Plc (“Carillion”). Carillion was a British construction company whose receivers petitioned for compulsory liquidation in January 2018 less than a year after the signing of its most recent annual report (Companies House, 2018). The company’s last ever annual report and financial statements dated 31 December 2016 were signed on March 2017 and were prepared on a going concern basis. The auditors report failed to highlight any concerns regarding the company’s health (ACCA, 2018).

The collapse of Carillion is just one example. Other corporate failures including BHS and UK Lender HBOS have raised concerns regarding why these companies were issued clean audit opinions. Auditors are being asked to be more vigorous when assessing the appropriateness of the going concern assumption (Financial Times, 2019).

### The Challenges of Auditing the Going Concern Assumption
Before continuing it is important to acknowledge the difficulties and inherit limitations auditors face when auditing the going concern assumption. Going concern is the prediction of a future event. ISAs acknowledge that the potential effects and limitations on the auditor’s ability to detect misstatements are greater for future events as the auditor cannot predict such events (IAASB, 2015). Therefore, we can not **entirely** blame auditors for failing to identify future corporate failures.

However, it is equally important to acknowledge that oftentimes, there are indications or evidence that the entity may, in the near future find itself in financial distress. The evidence obtained by the auditor to allow them to conclude on the appropriateness of the going concern assumption can contain many elements. The guidance outlined by ISAs outlines an array of procedures an auditor shall perform when fulfilling their responsibilities relating to going concern. However, in this paper, we will focus one key area of this guidance. Namely, International Standard on Auditing 570 paragraph 10 (ISA 570.10). (IAASB, 2015) ISA 570.10 states:
> “When performing risk assessment procedures […] the auditor shall consider whether events or conditions exist that may cast significant doubt on the entity’s ability to continue as a going concern […]. The following are examples of events or conditions that, individually or collectively, may cast significant doubt on the entity’s ability to continue as a going concern […]

>- Financial
    - Net liability or net current liability position
    - Fixed-term borrowings approaching maturity without realistic prospects of renewal or repayment; or excessive reliance on short-term borrowings to finance long-term assets.
    - […]
    - Negative operating cash flows.
    - Adverse key financial ratios.
    - Substantial operating losses or significant deterioration in the value of assets used to generate cash flows.
    - Arrears or discontinuance of dividends.
    - Inability to comply with the terms of loan agreements.”
>- ISA 570.10 outlines further financial and non-financial considerations an auditor may make in identifying conditions that cast doubt on an entity’s going concern assumption however these are beyond the scope of this paper.

Circumstances may arise where one of the above conditions is fulfilled and the significant doubt on the entity’s ability to continue as a going concern is self-evident. For example, an entity’s looming debt obligations with no liquidity to settle them will cast obvious doubt on its ability to continue as a going concern. Difficulty arises when no single condition or event when considered individually casts significant doubt on going concern and the auditor is required to evaluate a collective of conditions (hereafter “ratios”). This may lead to the following issues:
- It may be difficult to identify exactly which financial statement ratios are most relevant for the decision.
- It is difficult to interpret each ratio in the context of the collective. For example, assuming it has been determined that a ratio is relevant to our decision:
    - How much weight (importance) will we give it?
    - Is the ratio’s value in relation to other ratios relevant? If the ratio is adverse, is this mitigated or exacerbated by other ratios?
    - Let’s assume that once the ratios are analysed, we can formulate a record of the company’s financial health. In order for us to interpret our record it may be most suitable to evaluate it in the context of the industry.

When we consider the points raised above, we see the some of the challenges of auditing an entity’s going concern consumption. The assessment is complex and many aspects of it are subjective in nature.

The auditor’s ability to reach an appropriate conclusion is further limited by their tool kit. Within auditing firms, spreadsheets are still the most dominant tool utilised to create workpapers (AICPA, 2021). Spreadsheets are extremely popular, they are easy to use and come preinstalled on most computers. Despite their popularity, spreadsheets have their limitations particularly when it comes to statistical analysis (IBM, 2018).

#### Measuring Performance
We use recall and AUC to measure performance. (Müller & Guido, 2016) Outline that recall measures how many positive samples are captured by positive predictions. We use recall as a performance metric when we need to identify all positive samples (i.e that a company shows indications of being bankrupt). This is certainly the case when auditing going concern. Accurately identifying bankruptcy cases is crucial not only for auditing firms in terms of reputational and financial (through fines and litigation) damage, but also for the users of the financial statements who may make economic decisions based on them.

There can be negative side effects to optimising our models for recall. (Müller & Guido, 2016) outline that precision measures how many of the cases our model predicted as positive were actually positive. It is possible for us to trivially obtain perfect recall if our model flags every instance as being a bankrupt which would lead to a high number of false flags (low precision). Regarding false flags:
- The best-case scenario for a false flag is that the auditor performs further analysis and in the precise of other information, concludes that there is no evidence that casts significant doubt on the going concern assumption.
- The worst-case scenario is that the auditor fully relies on the model's false flag prediction and erroneously increases the scope of the audit due to the perceived risk. As a result, more resources will be allocated to address the risk. Although this may decrease the profitability of an audit, it is still preferred to the scenario where our ability to identify bankruptcies is low.

Nonetheless if we want to instil confidence in our model it is important that our model has acceptable levels of precision and recall. Correspondingly, we use Area Under Curve (AUC) as a second performance metric. The AUC will measure the aggregate performance of precision and recall (Müller & Guido, 2016).

#### Addressing Poor Performance with Up-sampling and Down-sampling
As noted above, the number of bankrupt cases in our original dataset represents only 3.226% of the total number of cases. Further down the pipeline we noticed that all models initially performed poorly. We believe this was due to the low proportion of bankruptcy cases in our dataset. Most likely, there are not enough bankruptcy instances to successful train our model.

The code block below allows us to down-sample and up-sample our dataset:
- When we down-sample we decrease the number of non-bankrupt instances in our dataset. This has the disadvantage of omitting information from our data set. Indeed, when we first performed this, we noticed that both AUC and recall increased however precision suffered.
- When we up-sample we increase the number of bankrupt instances. The advantage to this resampling method is that we do not lose information. The disadvantage of this resampling method is that by increase the number of duplicate instances in our data which will lead to overfitting. As such we limit the number of duplications.

We can toggle resampling setting the Boolean flags downsample_flag and upsample_flag in the first code block of this notebook. After careful analysis, we up-sampling is the preferred approach to address class imbalance. The up-sampling resulted in increased AUC and recall without adversely affecting precision.

```python
#Downsample
#Code template inspired by https://elitedatascience.com/imbalanced-classes
from sklearn.utils import resample
upsample_multiplier = 2
minority = sum(rawdata["Bankrupt?"]==1)
# Separate majority and minority classes
data_majority = rawdata[rawdata["Bankrupt?"]==0]
data_minority = rawdata[rawdata["Bankrupt?"]==1]
data_majority_downsampled = resample(data_majority, 
                                     replace=False,    # sample without replacement
                                     n_samples=220,     # to match minority class
                                     random_state=seed) # reproducible results

# Combine minority class with downsampled majority class
data_downsampled = pd.concat([data_majority_downsampled, data_minority])

data_minority_upsampled = resample(data_minority, 
                                     replace=True,     # sample with replacement
                                     n_samples=(minority*upsample_multiplier),    # trial and error to optimise
                                     random_state=seed) # reproducible results

data_upsampled = pd.concat([data_majority, data_minority_upsampled])

if downsample_flag:
    rawdata = data_downsampled
    
if upsample_flag:
    rawdata = data_upsampled
    
summarise(rawdata)
```

## 2. Constructing and Selecting Features 
### Feature Selection with sklearn.ensemble.RandomForestClassifier and Professional Judgment
Our feature selection preprocessing consists of two parts:
- First, we will perform automatic preprocessing using RandomForestClassifer (RFC). We use RFC to evaluate the importance of each feature. We then rank the features by importance and select the 15 most important features. The author has deemed 15 to be an appropriate number through experimentation. If less than 15 features are selected, model performance could suffer as relevant features could be omitted by RFC. Conversely, when more than 15 features are selected the resulting list has many redundant features. 
- Finally, we use professional judgement to prune the list of 15 features. We eliminate features by gaining an understanding of each financial statement ratio and its purposes. We then reduce the number of features that provide overlapping information. For example the liability to equity ratio and the equity to liability ratios are the inverse of each other. 

In addition to RVC the author experimented with feature selection using sklearn.feature_selection.SelectKBest. In comparison to those of RFC, the features return by SelectKBest were less relevant and model performance suffered as a result. Feature selection with SelectKBest could be improved through future hyperparameter optimisation. However, at present, the ease of implementing accurate feature selection has been the deciding factor for selecting RFC.

## 2. Constructing and Selecting Features 
### Automatic Feature Selection with sklearn.ensemble.RandomForestClassifier

```python
#Initialise model and fit it
model = RandomForestClassifier(random_state=seed)
model.fit(X_scaled, Y)

#Initialise a list to rtack the importance scores of the features
ranks = [x for x in model.feature_importances_]
ranks.sort(reverse=True) #sort in descending order
threshold = ranks[15] #get the 16th highest value, this is the theshold for inclusion.

#Initialise a new list of the importances
importances = [x for x in model.feature_importances_]
#Populate the indexes of our features providing they are above the threshold
RFC_mask = [True if x>threshold else False for x in importances]  
RFC_indexes = []
for i in range(len(RFC_mask)):
    if RFC_mask[i]==True:
        RFC_indexes.append(i)
```

```python
#exclude redundant features

#maintain a list of redundant features
redundant_features = ["ROA(A) before interest and % after tax",
                      "Net Value Per Share (A)",
                      "Net Value Per Share (C)",
                      "ROA(B) before interest and depreciation after tax",
                      "Fixed Assets Turnover Frequency",
                      "Revenue per person",
                      "Fixed Assets to Assets",
                      "Equity to Liability",
                      "Persistent EPS in the Last Four Seasons",
                      " Net worth/Assets"
                     ]
# initialise mask for the final features of our training data
final_features = []

#finalise the mask representing the final features of our training data
for col in RFC_indexes:
    ratio = headers[col+1].strip()
    if ratio not in redundant_features:
        final_features.append(col+1)

        
final_features = np.append([0],final_features) # add back our target variable to the mask
```

## 3. Building our Machine Learning Models
We can now begin building our models. We create a pipeline for several models and then print a classification report. We see that the tree classifiers are the best performing. Our research confirms that utilising tree classifiers is a common way addressing class imbalance in data sets (Brownlee, 2020).

We further note that the literature suggests that tree classifiers are indifferent to feature scaling (Zeng & Casari, 2016) however in our runs we noticed that their performance improved slightly when the features were unscaled. For this reason, the tree classifiers are trained with unscaled data.

```python
#split and normalise
test_size = 0.33

X_train, X_test, Y_train, Y_test = train_test_split(X_extracted, Y_extracted, test_size=test_size, random_state=seed)

#Fit and transform features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# prepare models

from sklearn.metrics import roc_curve
from sklearn.metrics import roc_auc_score

models = []
models.append(((False,'LogisticRegression'), LogisticRegression(solver='liblinear')))
models.append(((False,'SVM'), SVC()))
models.append(((True,'DecisionTree'), DecisionTreeClassifier(random_state = seed))) 
models.append(((True,'RandomForest'),RandomForestClassifier(random_state = seed,
                                                           # n_estimators=100, 
                                                           # criterion = "gini"
                                                           )))
models.append(((False,'LDA'), LinearDiscriminantAnalysis()))
models.append(((False,'NB'), GaussianNB()))


#Run reports
for name, model in models:
    if name[0]:
        model.fit(X_train, Y_train) #noticed that the tree calssifiers performed better on non scaled data
        predicted = model.predict(X_test)
        report = classification_report(Y_test, predicted)
        print(model)
        print(report)
        print("The AUC is of",name[1],"is:","{0:.2%}".format(roc_auc_score(Y_test, predicted)))
        print()
    else:
        model.fit(X_train_scaled, Y_train)
        predicted = model.predict(X_test_scaled)
        report = classification_report(Y_test, predicted)
        print(model)
        print(report)
        print("The AUC is of",name[1],"is:","{0:.2%}".format(roc_auc_score(Y_test, predicted)))
        print()

    
```
